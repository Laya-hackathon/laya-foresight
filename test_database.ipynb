```python
import os
import pandas as pd
from sqlalchemy import create_engine

with open(".env") as f:
    for line in f:
        if "=" in line:
            k, v = line.strip().split("=", 1)
            os.environ[k] = v

engine = create_engine(
    os.environ["DATABASE_URL"],
    connect_args={"connect_timeout": 10}
)


with engine.connect() as conn:
    print("Connected!")
```

```python
pd.read_sql("""
SELECT table_name 
FROM information_schema.tables 
WHERE table_schema = 'public';
""", engine)
```

```python
try:
    df = pd.read_sql("SELECT * FROM model_predictions", engine)
    print(df.head())  # Display the first few rows of the DataFrame
    print("Data fetched from database successfully.")
except Exception as e:
    print(f"Failed to connect to database: {e}")
    print("Loading data from local CSV file instead.")
    # Assuming the CSV is in the data directory relative to the script
```
